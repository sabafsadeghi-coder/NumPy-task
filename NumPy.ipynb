```python
# 2- Python list vs NumPy Array :

import numpy as np

# Python List
my_list = [20, 40, 60, 80, 100]

# NumPy Array
my_array = np.array([20, 40, 60, 80, 100])

print("Python List:", my_list)
print("NumPy Array:", my_array)
```

```python
# 4- Important Array Properties :

import numpy as np

data = np.array([[2, 4, 6],
                 [20, 40, 60]])

print("Shape:", data.shape)
print("Size:", data.size)
print("Number of dimensions:", data.ndim)
print("Data type:", data.dtype)
```

```python
# 5- Indexing :

import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

print(numbers[-2])
print(numbers[0])
print(numbers[2])
```

```python
# 6-Indexing in a 2D Arrays:

import numpy as np

data = np.array([[2, 4, 6],
                 [20, 40, 60]])

print(data[0, 0])
print(data[0, 2])
print(data[1, 1])
```

```python
# 7- Slicing :

import numpy as np

numbers = np.array([20, 40, 60, 80, 100])

print(numbers[1:4])
```

```python
# 8- Vectorized Operations :

import numpy as np

numbers = np.array([20, 40, 60, 80, 100])

print(numbers + 5)
print(numbers * 2)
print(numbers - 5)
print(numbers / 2)
```

```python
# 9- Operations Between Two Array :

import numpy as np

a = np.array([20, 40, 60])
b = np.array([2, 4, 6])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)

```

```python
# 10- Mathematical Functions :

import numpy as np

numbers = np.array([9, 16, 49, 81])

print("Square root:", np.sqrt(numbers))
print("Absolute value:", np.abs(numbers))
print("Sum:", np.sum(numbers))
print("Maximum:", np.max(numbers))
print("Minimum:", np.min(numbers))
```

```python
# 11- Aggregation Functions :

import numpy as np

marks = np.array([20, 40, 60, 80, 100])

print("Sum:", np.sum(marks))
print("Mean:", np.mean(marks))
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
print("Median:", np.median(marks))
print("Standard Deviation:", np.std(marks))
```

```python
# 12- Conditions in NumPy :

import numpy as np

marks = np.array([20, 40, 60, 80, 100])

result = marks >= 50

print(result)
```

```python
# 13- Filtering Data :

import numpy as np

marks = np.array([20, 40, 60, 80, 100, 110, 120])

filtered_marks = marks[marks >= 50]

print("Filtered marks:", filtered_marks)
```

```python
# 14- Reshape :

import numpy as np

numbers = np.array([2, 4, 6, 8, 10, 12])

new_array = numbers.reshape(2, 3)

print(new_array)
```

```python
# 15- Flatten :

import numpy as np

data = np.array([[2, 4, 6],
                 [8, 10, 12]])

flat_data = data.flatten()

print("Original array:")
print(data)

print("Flattened array:")
print(flat_data)
```
